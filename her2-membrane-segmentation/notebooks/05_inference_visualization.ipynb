# Inference Visualization

This notebook allows interactive visualization of predictions from all models.

```
import matplotlib.pyplot as plt
import os
from PIL import Image

models = ['unet', 'unet_plusplus', 'attention_unet', 'nnunet', 'segformer', 'unetformer', 'swin_unet']
sample_idx = 0

plt.figure(figsize=(20, 10))
for i, model in enumerate(models):
    pred_path = f'../outputs/{model}/predictions/pred_{sample_idx}.png'
    if os.path.exists(pred_path):
        img = Image.open(pred_path)
        plt.subplot(2, 4, i+1)
        plt.imshow(img)
        plt.title(model)
        plt.axis('off')
plt.tight_layout()
plt.show()
```
