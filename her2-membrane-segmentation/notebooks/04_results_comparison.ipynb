# Results Comparison

This notebook pulls the final evaluation metrics from all models and creates comparison tables and plots.

```
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
import os

results_path = '../outputs/comparison_results.csv'
if os.path.exists(results_path):
    df = pd.read_csv(results_path)
    display(df)
    
    # Plotting
    df_melted = df.melt(id_vars=['model_name'], value_vars=['dsc', 'iou', 'f1', 'hd95'], var_name='Metric', value_name='Score')
    plt.figure(figsize=(12, 6))
    sns.barplot(data=df_melted, x='model_name', y='Score', hue='Metric')
    plt.title('Model Performance Comparison')
    plt.xticks(rotation=45)
    plt.tight_layout()
    plt.show()
else:
    print('Run scripts/evaluate_all.py first to generate the comparison results.')
```
